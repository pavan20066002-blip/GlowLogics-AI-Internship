1️⃣ Calculate the Average Marks

* Compute the **average marks** of all students in the dataset.

```python
import pandas as pd
df = pd.read_csv('/content/student_performance_dataset.csv')
avg_marks = df['Marks'].mean()
print(f"Average marks: {avg_marks:.2f}")
```

2️⃣ High Performers Analysis

* Filter the dataset to determine **how many students scored above 75 marks**.

```python
high_performance = df[df['Marks'] > 75]
high_performance_count = len(high_performance)
print(f"Number of students with marks above 75: {high_performance_count}")
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['Attendance_Percentage'], df['Marks'])
plt.xlabel('Attendance_Percentage')
plt.ylabel('Marks')
plt.title('Marks vs Attendance_Percentage')
plt.grid(True)
```

3️⃣ Correlation Analysis

* Analyze whether **higher attendance leads to better marks**.
* Calculate the **correlation coefficient** between Attendance and Marks.
* Briefly interpret the result.

```python
correlation = df['Attendance_Percentage'].corr(df['Marks'])
print(f"Correlation between Attendance_Percentage and Marks: {correlation:.2f}")
```

4️⃣ Feature Engineering

* Create a new column called **`Pass`** where:
Pass = True  if Marks >= 50  
Pass = False if Marks < 50

```python
df['Pass'] = df['Marks'] >= 50
```

```python
df['Pass'] = df['Marks'].apply(lambda x: True if x >= 50 else False)
```

5️⃣ Summary Statistics

* Count the total number of:

  * ✅ Passes
  * ❌ Fails

6️⃣ Visualization

* Create a **distribution plot (histogram)** of student marks.
* Add appropriate:

  * Title
  * Axis labels
  * Grid (optional)

```python
passes = df['Pass'].sum()
fails = (~df['Pass']).sum()
print(f"Passes: {passes}")
print(f"Fails: {fails}")
plt.figure(figsize=(8,5))
plt.hist(df['Marks'], bins=6, edgecolor='black', color='skyblue')
plt.xlabel('Marks')
plt.ylabel('Number of students')
plt.title('Distribution of Marks')
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.show()
```
