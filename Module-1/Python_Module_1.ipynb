**if-else statement**

```python
age = 20
if age > 18:
  print("You can vote")
else:
  print("You cannot vote")
```

```python
age = int(input("Enter your age: "))
if age > 18:
  print("You can vote")
else:
  print("You cannot vote")
```

```python
light = "green"
if light == ("green"):
  print("you can drive a car")
else:
  print("stop at red signal")
```

**elif Statement**

```python
marks = 70
if marks >= 90:
  print("Grade A")
elif marks >= 80:
  print("Grade B")
elif marks >= 70:
  print("Grade C")
```

```python
marks = int(input("Enter your marks: "))
if marks >= 80:
  print("Grade A")
elif marks >= 70:
  print("Grade B")
elif marks >= 60:
  print("Grade C")
else:
  print("Grade D")
```
